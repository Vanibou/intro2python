````python
#### Week 4 

# While loops are used to repeat a block of code as long as a condition is true. The syntax is:
pets = 0
while pets<25:
  print('Marcel demands pets! You have given him '+str(pets)+ ' pets.')
  pets +=1
print('Marcel is satisfied after receiving '+str(pets) + ' pets.')

# Collatz sequence:The Collatz sequence begins with some starting number. Given its previous number $n$, it generates the next number based on the following rule: if $n$ is even, the next number is $n/2$. If $n$ is odd, the next number is $3n+1$.
# Write some code below that generates the Collatz sequence for a starting number of $27$ until the number hits $1$.
n = 27
collatz = [n]
while n != 1:       # Run loop while n is not 1  
    if n % 2 == 0:  # if even
        n /= 2      # Divide by two
    else:           # if odd
        n = 3*n + 1 
    collatz.append(n)
print(collatz)


## Break and continue statements
#When code reaches a `continue` statement, it jumps to the next iteration of the loop without running the rest of the code in the loop.
for i in range(10):
  print("loop #: ", i)
  if i%2==0:
    continue
  print(i) # This command is skipped for all even numbers

# When code reaches a `break` statement it jumps outside the loop (skipping any remaining iterations) without running the rest of the code in the loop.
for i in range(10):
  print("loop #: ", i)
  if i==5:
    break # Leaves the loop completely once it encounters this statement
  print(i)

## Try and excepted statements
# The `try` and `except` statements are used to catch errors in code. If an error occurs in the `try` block, the code in the `except` block will run

word = "abracadabra"

letter_counts = {}
for letter in word:
  try:
    letter_counts[letter]+=1
  except KeyError: # we specify the type of error we expect here
    letter_counts[letter] = 1

print(letter_counts)


## Functions 
def fib_return (n):
  """
  Print fibonnacci series up to n.
  Args:
      n (int): Maximum value of the fibonacci series to return.
  Returns:
      list: List of fibonacci sequence values
  """

  fib_list = []
  a,b = 0, 1
  while a <n:
    fib_list.append(a)
    next= a+b
    a = b
    b = next
  return fib_list

def f(a, L=None):
    if L is None: # None now provides a marker that we would like to use the default argument.
        L = []
    L.append(a)
    return L

print(f(1))
print(f(2))
print(f(3)) -> [1], [2], [3]



def collatz(start = 100, end = 1):
    """Function that prints the collatz sequence
    --- inputs ---
    start, (int): starting point
    end,   (int): end point
    """
    n = start
    collatz = [n]
    while n != end:      
        if n % 2 == 0:  # if even
            n /= 2      # Divide by two
        else:           # if odd
            n = 3*n + 1 
        collatz.append(n)
    return collatz


### F strings 
price = 59
txt = f"The price is {price:.2f} dollars"  # : means format the placeholder, and .2f means format as a float with 2 decimal places
print(txt)

price = 59
txt = f"The price is {2 * price + 10} dollars"  # : means format the placeholder, and .2f means format as a float with 2 decimal places
print(txt)
````
