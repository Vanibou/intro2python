## 3. Indexing: columns, `.loc`, `.iloc`

**Reminder**
- `df['col']` → one column
- `df[['col1', 'col2']]` → several columns (double brackets = a list)
- `df.loc[row, 'col']` → by row number and column **name**
- `df.iloc[row, col]` → by row number and column **number**

```python
print(df_p['name'])
print(df_p[['name', 'age']])
print(df_p.loc[1, 'name'])     # row 1, column 'name'
print(df_p.iloc[1, 0])         # row 1, column number 0 (= 'name')
```

## 7. Dot-motion data: subsetting and `accuracy`

In the real notebook you read `dot_motion.csv`. Since I don't have that file here, the next cell **simulates a very similar file** (same columns: `subject`, `motion`, `noise_level`, `choice`, `reaction_time`) and saves it as `dot_motion.csv`. If you already have the real file in this folder, **skip the next cell** and just run the `read_csv`.

In this task, people see moving dots and decide whether they move **left** or **right**. `motion` is the true direction, `choice` is what the participant answered.

```python
# --- Simulate a dot_motion.csv (skip if you have the real one) ---
rng = np.random.default_rng(1)
n = 120
sim = pd.DataFrame({
    'subject': np.repeat(['Subject 1', 'Subject 2', 'Subject 3'], n // 3),
    'motion': rng.choice(['left', 'right'], n),
    'noise_level': rng.choice(['low', 'high'], n),
})
p_correct = np.where(sim['noise_level'] == 'low', 0.9, 0.7)
correct = rng.random(n) < p_correct
other = np.where(sim['motion'] == 'left', 'right', 'left')
sim['choice'] = np.where(correct, sim['motion'], other)
sim['reaction_time'] = rng.gamma(shape=20, scale=28, size=n) + 50
sim.to_csv('dot_motion.csv')        # index saved on purpose, like the original file
```

## Quick recap
| I want to... | Code |
|---|---|
| Make a data frame from a list of dicts | `pd.DataFrame(list_of_dicts)` |
| Rows × columns | `df.shape` |
| One / several columns | `df['a']`, `df[['a', 'b']]` |
| By row number + column name | `df.loc[1, 'a']` |
| By row number + column number | `df.iloc[1, 0]` |
| Keep rows that meet a condition | `df[df['a'] > 5]` |
| Combine conditions | `df[(cond1) & (cond2)]` |
| Add a column | `df['new'] = ...` |
| Save / load | `df.to_csv('f.csv', index=False)`, `pd.read_csv('f.csv')` |
| 1/0 "do these match?" | `(df['a'] == df['b']).astype(int)` |
